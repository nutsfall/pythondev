## 仕様

- 空のリストを作成して英単語を格納する  
- キーが a〜z、値がすべて0の辞書を作成する  

- キーボードから英単語を入力する  
  - 未入力でエンターが押されるまで繰り返す  
  - 入力された単語はリストに追加する  
  - 単語の各文字を1文字ずつ確認し、対応する辞書の値を加算する  

- 最後に結果を表示する  
  - 入力された英単語のリストをABC順に並べ替えて表示する  
  - 出現回数が1以上の文字について「○が△個ありました」と表示する  

```python
alphabet = dict.fromkeys((chr(i) for i in range(ord('a'), ord('z') + 1)), 0)

inputed_words = []
while True:
    word = input("英単語を入力してください： ").lower()
    if word == "":
        break

    inputed_words.append(word)
    for char in word:
        if char in alphabet:
            alphabet[char] += 1

print(f"入力した英単語：{', '.join(inputed_words)}")

for char, count in sorted(alphabet.items()):
    if count > 0:
        print(f"{char}が{count}個ありました") 
```
