## 概要

以下は架空の野球リーグの勝敗表です。

| チーム名 | 勝利 | 敗北 | 引分 |
|----------|------|------|------|
| Archers  | 77   | 64   | 2    |
| Meteors  | 71   | 69   | 3    |
| Cheetahs | 69   | 68   | 6    |
| Sharks   | 70   | 70   | 3    |
| Pythons  | 68   | 73   | 2    |
| Owls     | 59   | 82   | 2    |

勝ち負け数を管理するクラスを作成し、6チーム分のインスタンスを生成して、各チームの勝敗情報を表示するプログラムを作成してください。

## 仕様

- 最初のセルで `Baseball_Team` クラスを作成する  
  - メンバ変数  
    - `name`（チーム名）  
    - `win`（勝ち数）  
    - `lose`（負け数）  
    - `draw`（引き分け数）  

- メソッド `calc_win_rate()` を作成する  
  - 勝率 = 勝ち数 / (勝ち数 + 負け数)  
  - 引き分けは考慮しない  
  - メンバ変数ではなくメソッドで算出する  

- メソッド `show_team_result()` を作成する  
  - 1チーム分の勝敗情報を表示する  
  - 例：`Archers 77 64 2 0.546`  

- 次のセルでインスタンスを6チーム分作成する  
  - 各チームの名前・勝ち数・負け数・引き分け数を設定する  

- 各インスタンスの `show_team_result()` を呼び出す  
  - 表の見出しも表示する  

```python
class Baseball_Team:
    def __init__(self, name, wins, lose, draw):
        self.name = name
        self.wins = wins
        self.lose = lose
        self.draw = draw

    def calc_win_rate(self):
        games_of_win_and_lose = self.wins + self.lose
        if games_of_win_and_lose == 0:
            return 0.0
        return self.wins / games_of_win_and_lose

    def show_team_result(self, row_format="{name} {win} {lose} {draw} {rate:.3f}"):
        win_rate = self.calc_win_rate()
        print(
            row_format.format(
                name=self.name,
                win=self.wins,
                lose=self.lose,
                draw=self.draw,
                rate=win_rate,
            )
        )

class Baseball_League:
    HEADER_FORMAT = "{name:<10} {win:>3} {lose:>4} {draw:>4} {rate:>6}"
    ROW_FORMAT = "{name:<10} {win:>3} {lose:>4} {draw:>4} {rate:>6.3f}"

    def __init__(self, teams):
        self.teams = teams
    
    @classmethod
    def get_header_text(cls):
        return cls.HEADER_FORMAT.format(name="team", win="win", lose="lose", draw="draw", rate="rate")

    def show_league_result(self):
        header = self.get_header_text()
        print(header)
        print("-" * len(header))
        for team in self.teams:
            team.show_team_result(self.ROW_FORMAT)
```

```python
Baseball_League([
    Baseball_Team("Archers", 77, 64, 2),
    Baseball_Team("Bandits", 71, 69, 3),
    Baseball_Team("Cheetahs", 69, 68, 6),
    Baseball_Team("Sharks", 70, 70, 3),
    Baseball_Team("Pythons", 68, 73, 2),
    Baseball_Team("Owls", 59, 82, 2)
]).show_league_result() 
```
