## 仕様

### prognum.py
- フィボナッチ数列を計算する関数を定義する（関数名・引数名は任意）
- ルール
  - 1番目 = 1
  - 2番目 = 1
  - 3番目以降 = 直前2つの和
    - 例:
      - 3番目: 1 + 1 = 2
      - 4番目: 1 + 2 = 3
      - 5番目: 2 + 3 = 5
      - 6番目: 3 + 5 = 8
- 再帰関数で実装する
- 無限ループにならないよう終了条件を設定する

- 計算結果を画面に出力する

```python
import prognum 

number = int(input("何番目の数値を求めますか？： "))
print(f"{number}番目のフィボナッチ数は {prognum.fibonacci(number)} です")
```
